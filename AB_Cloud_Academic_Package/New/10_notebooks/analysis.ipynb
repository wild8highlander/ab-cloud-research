# Analysis — the decisive quantities

The four quantities that carry the statistical layer, on their pre-registered scales.

```python
metrics = {
    'GUE r (zeta-decorated)': (0.608, 0.5, 0.7),
    'GUE r (theory)': (0.5992, 0.5, 0.7),
    'trace-primes r': (0.9996, 0.9, 1.0),
    'locality wall (S_A/190)': (14/190, 0.0, 1.0),
}
for k, (v, lo, hi) in metrics.items():
    x = max(0.0, min(1.0, (v - lo) / (hi - lo)))
    print(f'{k:<28}', '#' * round(x * 40), f'{v}')
```
